```python
# Practical 4: Normalization, standardization, binning and encoding
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, LabelEncoder
 
df = pd.DataFrame({
    "Age": [22, 25, 30, 35, 40, 45, 52, 58],
    "Salary": [25000, 32000, 45000, 52000, 61000, 75000, 88000, 95000],
    "City": ["Pune", "Mumbai", "Pune", "Nashik",
             "Mumbai", "Nashik", "Pune", "Mumbai"],
    "Education": ["Graduate", "Masters", "Graduate", "PhD",
                  "Masters", "Graduate", "PhD", "Masters"],
})
print("Original data:\n", df)
 
# 1. Normalization (Min-Max): brings values between 0 and 1
df["Salary_Norm"] = MinMaxScaler().fit_transform(df[["Salary"]]).round(3)
 
# 2. Standardization (Z-score): mean = 0 and standard deviation = 1
df["Salary_Std"] = StandardScaler().fit_transform(df[["Salary"]]).round(3)
 
print("\nNormalization and Standardization:")
print(df[["Salary", "Salary_Norm", "Salary_Std"]])
print("Mean of standardized salary:", abs(round(df["Salary_Std"].mean(), 3)))
 
# 3. Binning: convert Age (numeric) into groups (categories)
df["Age_Group"] = pd.cut(df["Age"], bins=[0, 30, 45, 100],
                         labels=["Young", "Middle-aged", "Senior"])
print("\nBinning of Age:\n", df[["Age", "Age_Group"]])
 
# 4. Label encoding: each category gets a number
le = LabelEncoder()
df["Education_Label"] = le.fit_transform(df["Education"])
print("\nLabel Encoding of Education:\n", df[["Education", "Education_Label"]])
print("Classes (in order of numbers):", list(le.classes_))
 
# 5. One-hot encoding: each category gets its own 0/1 column
onehot = pd.get_dummies(df["City"], prefix="City", dtype=int)
print("\nOne-Hot Encoding of City:\n", pd.concat([df["City"], onehot], axis=1))
```
