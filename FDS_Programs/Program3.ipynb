```python
# Practical 3: Duplicates, inconsistent values and missing data
import pandas as pd
import numpy as np
 
data = {
    "ID": [1, 2, 3, 4, 4, 5, 6, 7, 8, 9],
    "Name": ["Amit", "Riya", "Karan", "Sneha", "Sneha",
             "Rahul", "Pooja", "Vikas", "Neha", "Arjun"],
    "City": ["Pune", "pune ", "MUMBAI", "Nashik", "Nashik",
             "Mumbai", None, "PUNE", "Nagpur", "Mumbai"],
    "Age": [20, 21, np.nan, 22, 22, 20, 23, np.nan, 19, 20],
    "Marks": [78, 85, 62, np.nan, np.nan, 71, 88, 55, np.nan, 93],
    "Fees": [50000, 52000, np.nan, 48000, 48000,
             np.nan, 50000, 51000, 49000, 52000],
}
df = pd.DataFrame(data)
print("Original data:\n", df)
 
# ---------- Step 1: Identify problems ----------
print("\nNumber of duplicate rows:", df.duplicated().sum())
print("\nMissing values in each column:\n", df.isnull().sum())
print("\nUnique values in City column:", list(df["City"].unique()))
 
# ---------- Step 2: Remove duplicates ----------
df = df.drop_duplicates()
print("\nShape after removing duplicates:", df.shape)
 
# ---------- Step 3: Fix inconsistent values ----------
df["City"] = df["City"].str.strip().str.title()   # remove spaces, same case
print("Unique values in City after cleaning:", list(df["City"].unique()))
 
# ---------- Step 4: Handle missing data ----------
# Method 1: Deletion (on a copy, only to see the effect)
print("\nRows left if we delete all rows with missing data:",
      len(df.dropna()), "out of", len(df))
 
# Method 2: Fill missing values
df["City"] = df["City"].fillna(df["City"].mode()[0])          # mode
df["Age"] = df["Age"].fillna(round(df["Age"].mean(), 1))      # mean
df["Marks"] = df["Marks"].fillna(df["Marks"].median())        # median
df["Fees"] = df["Fees"].ffill()                               # forward fill
 
print("\nCleaned data:\n", df)
print("\nMissing values now:", df.isnull().sum().sum())
```
