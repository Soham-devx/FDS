```python
# Practical 5: Descriptive statistics, frequency distribution, correlation
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
 
# Load the Iris dataset (comes with scikit-learn)
iris = load_iris(as_frame=True)
df = iris.frame
df["species"] = df["target"].map(dict(enumerate(iris.target_names)))
df = df.drop(columns="target")
df.columns = ["sepal_length", "sepal_width", "petal_length",
              "petal_width", "species"]
print("First 5 rows:\n", df.head())
print("Shape:", df.shape)
 
# ---------- 1. Descriptive statistics ----------
num = df.drop(columns="species")          # only numeric columns
stats = pd.DataFrame({
    "Mean": num.mean(),
    "Median": num.median(),
    "Mode": num.mode().iloc[0],
    "Std Dev": num.std(),
    "Variance": num.var(),
    "Skewness": num.skew(),
}).round(2)
print("\nDescriptive statistics:\n", stats)
 
# ---------- 2. Frequency distribution ----------
print("\nFrequency of each species:\n", df["species"].value_counts())
bins = pd.cut(df["petal_length"], bins=[0, 2, 4, 5, 7])
print("\nFrequency distribution of petal length:")
print(bins.value_counts().sort_index())
 
# ---------- 3. Correlation analysis ----------
corr = num.corr().round(2)
print("\nCorrelation matrix:\n", corr)
 
# ---------- 4. Graphical representation ----------
fig, ax = plt.subplots(2, 2, figsize=(11, 8))
sns.histplot(df["sepal_length"], kde=True, ax=ax[0, 0])
ax[0, 0].set_title("Histogram of Sepal Length")
sns.heatmap(corr, annot=True, cmap="coolwarm", ax=ax[0, 1])
ax[0, 1].set_title("Correlation Heatmap")
sns.scatterplot(data=df, x="petal_length", y="petal_width",
                hue="species", ax=ax[1, 0])
ax[1, 0].set_title("Petal Length vs Petal Width")
sns.boxplot(data=df, x="species", y="petal_length", ax=ax[1, 1])
ax[1, 1].set_title("Petal Length of each Species")
plt.tight_layout()
plt.show()
```
