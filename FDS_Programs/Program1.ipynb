```python
# Practical 1: Variables, lists, tuples, dictionaries, functions and NumPy
import numpy as np
 
# ---------- 1. Variables and basic arithmetic ----------
a = 25
b = 4
print("a =", a, " b =", b)
print("Addition        :", a + b)
print("Subtraction     :", a - b)
print("Multiplication  :", a * b)
print("Division        :", a / b)
print("Floor division  :", a // b)
print("Remainder       :", a % b)
print("Power           :", a ** b)
 
# ---------- 2. List ----------
marks = [78, 85, 62, 90, 71]
marks.append(88)              # add a new element
marks.sort()                  # sort the list
print("\nSorted marks     :", marks)
print("Highest / Lowest :", max(marks), "/", min(marks))
print("Average marks    :", round(sum(marks) / len(marks), 2))
 
# ---------- 3. Tuple (cannot be changed) ----------
point = (10, 20, 30)
print("\nTuple            :", point)
print("First element    :", point[0])
print("Length of tuple  :", len(point))
 
# ---------- 4. Dictionary ----------
student = {"name": "Amit", "roll_no": 101, "course": "Data Science"}
student["grade"] = "A"        # add a new key
print("\nStudent details:")
for key, value in student.items():
    print("  ", key, "->", value)
 
# ---------- 5. Functions ----------
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
 
def is_even(n):
    return n % 2 == 0
 
print("\nFactorial of 5   :", factorial(5))
print("Is 14 even?      :", is_even(14))
 
# ---------- 6. NumPy arrays ----------
arr = np.array([10, 20, 30, 40, 50])
print("\nArray            :", arr)
print("Sum / Mean / Std :", arr.sum(), "/", arr.mean(), "/",
      round(arr.std(), 2))
print("Array + 5        :", arr + 5)
print("Array * 2        :", arr * 2)
print("Square root      :", np.sqrt(arr).round(2))
 
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print("\nMatrix A:\n", A)
print("Matrix B:\n", B)
print("A + B:\n", A + B)
print("A x B (matrix multiplication):\n", np.dot(A, B))
print("Transpose of A:\n", A.T)
print("Shape of A       :", A.shape)
print("Reshape 1-D to 2x3:\n", np.arange(1, 7).reshape(2, 3))
```
